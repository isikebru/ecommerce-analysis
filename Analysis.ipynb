# E-Commerce Sales Analysis

## Project Description
This project analyzes e-commerce sales data using Python.

## Goals
- Calculate total revenue
- Find best-selling products
- Analyze product performance

```python
orders = [
    {"product": "laptop", "price": 1200, "quantity": 1},
    {"product": "phone", "price": 800, "quantity": 2},
    {"product": "tablet", "price": 400, "quantity": 3},
    {"product": "phone", "price": 800, "quantity": 1},
    {"product": "laptop", "price": 1200, "quantity": 2},
]
```

```python
total_revenue = 0
for order in orders:
    
    total_revenue = total_revenue + order["price"] * order["quantity"]
print (total_revenue)
```

```python
for order in orders:
    revenue = order["price"] * order["quantity"]
    print (revenue)
```

```python
max_revenue = 0 
best_order = None
```

```python
for order in orders:
    revenue =order["price"] * order["quantity"]

    if revenue > max_revenue:
        max_revenue  = revenue 
        best_order = order

print(max_revenue)
print(best_order)
```

```python
revenue_by_product = {}
for order in orders:
    product = order["product"]
    revenue = order["price"] * order["quantity"]
    print(product, revenue)
```

```python
revenue_by_product = {}
for order in orders:
    product = order["product"]
    revenue = order["price"] * order["quantity"]
    if product in revenue_by_product:
        revenue_by_product[product] += revenue
    else:
        revenue_by_product[product] = revenue
print(revenue_by_product)
```

```python
max_revenue = 0
best_product = ""
for product, revenue in revenue_by_product.items():
    if revenue > max_revenue:
        max_revenue = revenue
        best_product = product
print (max_revenue, best_product) 
```
